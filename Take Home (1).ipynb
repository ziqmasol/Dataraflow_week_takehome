```python
# 1. **JSON File Handling**  
#    Write a program that stores a list of student records (name, age, grade) into a JSON file, and then reads the file back into a Python object. 
import json

# 1. Define the student records as a list of dictionaries
student_data = [
    {"name": "Salam", "age": 23, "grade": "45th grade"},
    {"name": "Alex", "age": 20, "grade": "A"}
]

with open("student_file.json", "w") as student_file:
    json.dump(student_data, student_file, indent=4)
    print("Data successfully saved to JSON file.\n")

with open("student_file.json", "r") as student_f:
    loaded_data = json.load(student_f)

print("Reading file back into Python:")
print(f"Data type: {type(loaded_data)}") # This will print <class 'list'>

for student in loaded_data:
    print(f"Name: {student['name']}, Age: {student['age']}, Grade: {student['grade']}")
```

```python
# 2. **Math Quiz**  Build a program that asks the user 5 random math questions (addition, subtraction, multiplication, division) and keeps score of correct answers.
import random

question_number = 5
score = 0  

while question_number > 0:
    random_int = random.randint(1, 20)  
    randomInt = random.randint(1, 20)
    
    operations = random.choice(["+", "-", "*", "/"])
    print(f"\nQuestion: {random_int} {operations} {randomInt}")
    
    answer = 0
    if operations == "+":
        answer = random_int + randomInt
    elif operations == "-":
        answer = random_int - randomInt
    elif operations == "*":
        answer = random_int * randomInt
    elif operations == "/":
        # FIX 2: Round division answers to 1 decimal place so it's fair
        answer = round(random_int / randomInt, 1)
        print("(Round your answer to 1 decimal place if needed)")

    # FIX 3: Use float() instead of int() so division answers don't crash the code
    input_ans = float(input("Input Answer: "))
    
    if input_ans == answer:
        score += 1
        print(f"Correct! Current Score: {score}")
    else:
        print(f"Wrong. The correct answer was {answer}. Current Score: {score}")
        
    question_number -= 1

print(f"\n🎉 Game Over! Your final score is {score}/5")

    
```

```python
# 4. **Date Calculator**  
#    Write a program that takes a user’s birthday as input and tells them how many days are left until their next birthday.  
from datetime import datetime, date

def next_birthday_Countdown(birthmonth,birthdate):
    current_date = date.today()

    next_birthday = date(current_date.year,birthmonth,birthdate)
    if next_birthday < current_date:
        next_birthday = date(current_date.year +1, birthmonth,birthdate)

    days_left = (next_birthday - current_date).days

    if days_left == 0:
        return "Happy birthday! it is day"
    else:
        return f"Hurray, you have {days_left} days left until your next birthday!"
    
print(next_birthday_Countdown(4,16))
```

```python
# 11. Write a program that calculates the number of days between two given dates. 
from datetime import date,datetime


def calculate_days(date1,date2):
    countdown_days = 0
    try:
        current_1 = datetime.strptime(date1, "%Y-%m-%d")
        current_2 = datetime.strptime(date2, "%Y-%m-%d")
    except:
        print("Enter a valid date")
    else:
        countdown_days = (current_2 - current_1).days
        print(countdown_days)
    
    return countdown_days


calculate_days("2025-2-20","2025-2-20")


print(datetime(2026,2,20) - datetime(2023,7,3))
```
